# [dbo].[usp_RunPayroll]

```sql
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

ALTER PROCEDURE dbo.usp_RunPayroll
(
    @Month INT,
    @Year INT
)
AS
BEGIN

    SET NOCOUNT ON;

    IF EXISTS
    (
        SELECT 1
        FROM PayrollRuns
        WHERE PayrollMonth = @Month
        AND PayrollYear = @Year
    )
    BEGIN
        THROW 50001,'Payroll already processed',1;
    END

    DECLARE @PayrollRunId INT;

    INSERT INTO PayrollRuns
    (
        PayrollMonth,
        PayrollYear
    )
    VALUES
    (
        @Month,
        @Year
    );

    SET @PayrollRunId = SCOPE_IDENTITY();

    DECLARE @WorkingDays INT = 26;

    INSERT INTO PayrollDetails
    (
        PayrollRunId,
        EmployeeId,
        BasicSalary,
        WorkingDays,
        DaysPresent,
        GrossPay,
        PFDeduction,
        ProfessionalTax,
        NetPay
    )
    SELECT
        @PayrollRunId,
        e.EmployeeId,
        e.BasicSalary,
        @WorkingDays,

        COUNT(a.AttendanceId) AS DaysPresent,

        ROUND(
            (e.BasicSalary / @WorkingDays)
            * COUNT(a.AttendanceId),
            2
        ) AS GrossPay,

        ROUND(
            (
                (e.BasicSalary / @WorkingDays)
                * COUNT(a.AttendanceId)
            ) * 0.12,
            2
        ) AS PFDeduction,

        200 AS ProfessionalTax,

        CASE
            WHEN COUNT(a.AttendanceId) = 0 THEN 0
            ELSE ROUND(
                (
                    (e.BasicSalary / @WorkingDays)
                    * COUNT(a.AttendanceId)
                )
                -
                (
                    (
                        (e.BasicSalary / @WorkingDays)
                        * COUNT(a.AttendanceId)
                    ) * 0.12
                )
                -
                200,
                2
            )
        END AS NetPay

    FROM Employees e

    LEFT JOIN Attendance a
        ON e.EmployeeId = a.EmployeeId
        AND MONTH(a.AttendanceDate) = @Month
        AND YEAR(a.AttendanceDate) = @Year
        AND a.IsPresent = 1

    WHERE e.IsActive = 1

    GROUP BY
        e.EmployeeId,
        e.BasicSalary;

END
GO
```
